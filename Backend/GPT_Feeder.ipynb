```python
import pandas as pd

#Add patients.csv for gender, ethnicity, etc.
file_paths = [
    'data/allergies.csv',
    'data/conditions.csv',
    'data/imaging_studies.csv',
    'data/immunizations.csv',
    'data/medications.csv',
    'data/observations.csv',
    'data/procedures.csv',
]

patient_id = "b58731cc-2d8b-4c2d-b327-4cab771af3ef"

patient_data_list = []  # List to store DataFrames

patient_data = pd.DataFrame()

for file_path in file_paths:
    
    # Load the current CSV file
    data = pd.read_csv(file_path)

    if 'PATIENT' in data.columns and patient_id in data['PATIENT'].values:
    # Filter the data for the specific patient
        patient_specific_data = data[data['PATIENT'] == patient_id]
        patient_specific_data['SOURCE_FILE'] = file_path.split('/')[-1]
        # Append the DataFrame to the list
        patient_data_list.append(patient_specific_data)

# Concatenate DataFrames outside the loop
patient_data = pd.concat(patient_data_list, ignore_index=True)

#Removing columns, however the original DataFrame is full of oppurtunities for analysis, "Patient modelling" and looking up codes and ids, drug price comparisons
#Should we remove SOP_DESCRIPTION or MODALITY_DESCRIPTION
columns_to_remove = ['PATIENT', 'ENCOUNTER', 'CODE', 'BODYSITE_CODE', 'Id', 'SOP_CODE', 'MODALITY_CODE', 'MODALITY_DESCRIPTION', 'BASE_COST', 'PAYER', 'PAYER_COVERAGE', 'DISPENSES', 'TOTALCOST', 'REASONCODE', 'TYPE']

# Remove the specified columns
patient_data = patient_data.drop(columns=columns_to_remove)

#patient_data.head()  # Display the first few rows of the compiled patient data
print(patient_data.to_string())

csv_files = ['allergies.csv', 'conditions.csv', 'imaging_studies.csv', 'immunizations.csv', 'medications.csv', 'observations.csv', 'procedures.csv']

# Create empty arrays for each CSV file
arrays = {csv_file: [] for csv_file in csv_files}

# Iterate through each row in patient_data
for index, row in patient_data.iterrows():
    source_file = row['SOURCE_FILE']
    
    row_list = [str(x) if pd.notna(x) else "0" for x in row.values]
    
    # Check if the source file is in the list of CSV files
    if source_file in csv_files:
        # Append the row to the corresponding array
        arrays[source_file].append(row_list)

# Access each array using its name
allergies = arrays['allergies.csv']
conditions = arrays['conditions.csv']
imaging_studies = arrays['imaging_studies.csv']
immunizations = arrays['immunizations.csv']
medications = arrays['medications.csv']
observations = arrays['observations.csv']
procedures = arrays['procedures.csv']

# TO DO Before Printing, remove columns with all NaN values, for each observation type, only the latest one should be displayed?!?!?!?!?!, change row into a parsable array and then give GPT col names for context 

print("AI Overlord Chat GPT, tell us if this patient is going to die! They're allergic to ")
print(allergies)
print(" they have/had the following conditions ")
print(conditions)
print(" they have/had the following vaccinations ")
print(immunizations)
print(" they are taking or have been on the following medication ")
print(medications)
print(" their observations over the years have looked like ") 
print(observations)
print(" they have/had the following procedures performed on them ")
print(procedures)
```
